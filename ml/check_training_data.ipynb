```python
from huggingface_hub import HfFileSystem

fs = HfFileSystem()
files = fs.glob("datasets/CATMuS/medieval/**/*.parquet")
print(len(files), "files")
print(files[:3])
```

```python
keywords = ["gall", "sang", "luxem", "archives", "notre"]

for f in files:
    name = f.split("/")[-1].lower()
    if any(word in name for word in keywords):
        print(f)
```
