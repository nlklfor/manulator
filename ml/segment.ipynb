```python
from PIL import Image

image = Image.open("data/page1.jpg")
image
```

```python
from kraken import blla
  
segmentation = blla.segment(image)
print("Number of lines:", len(segmentation.lines))
```

```python
from PIL import ImageDraw

drawing = image.copy()
pen = ImageDraw.Draw(drawing)

for line in segmentation.lines:
    boundary = [tuple(point) for point in line.boundary]
    baseline = [tuple(point) for point in line.baseline]
    pen.polygon(boundary, outline="red", width=3)
    pen.line(baseline, fill="blue", width=3)

drawing
```
